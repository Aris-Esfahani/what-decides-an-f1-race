Part 4: Conclusions

4.1 What actually decides a race?

Mostly where a driver starts, and the car that got them there.

- **The grid comes first.** Finishing order has always followed starting order, and more closely than ever: the grid-to-finish link rose from 0.58–0.65 before the 1990s to 0.70–0.77 since (1.4). In the model, scrambling grid position adds about 2 places of error per driver, more than twice as much as any other feature (3.5).
- **The car comes second.** Since 2000, the top team of each decade has won 47–49% of that decade's races (1.3), and the team's recent form is the model's second most useful feature (+0.74 places). The grid itself already reflects the car, since qualifying sets it.
- **The driver matters, but less than the car.** Once the grid and the team are known, a driver's own recent form adds only a little (+0.15 places). Drivers do differ: Verstappen turned 78% of his poles into wins, Hamilton and Schumacher about 58% (1.1).
- **Reliability used to decide races; now it rarely does.** Almost half of all starters retired in the 1950s, against 12% in the 2020s (1.2). With fewer cars breaking down, fewer drivers move up just by keeping going, which helps explain why the grid matters more now.
- **The popular stories barely matter.** Grid position isn't more important at Monaco (2.1), pit stops haven't gotten faster overall (2.2), the home crowd helps in qualifying but not in the race (2.3), and track specialists exist but the effect is small (2.4). In the model, street tracks, home races and track history add nothing (3.5).

Where the answer runs out is the front of the race. Using grid, team form and driver form, the model predicts the finishing order better than the grid alone (2.28 places off against 2.49, and a shuffle test says that isn't luck). But almost all of that gain comes from drivers starting 7th or lower. For the top 6 the model is no better than the grid: its pick won 56 of the 89 races in 2023–2026, against 54 for the pole sitter. Who wins among the front-runners probably comes down to things this data doesn't capture, like race strategy, tyres, safety cars and incidents.

4.2 Limitations

- **Retirements are left out of the model.** It predicts where a driver finishes if they finish. In 2023–2026 about 13% of starters didn't finish, and the model can't say who.
- **Car and driver can't be pulled apart.** The grid, team form and driver form all mix the two, so "the car comes second" is a reading of the results, not a measurement. 2.4 had the same problem with track specialists.
- **Data gaps.** Pit-stop data only covers 2011–2026 and measures the whole trip through the pit lane, not the tyre change (2.2). Lap times only go back to 1996 and weren't used. There's no weather, tyre or safety car data.
- **Rule changes.** The sport changes its rules every few years, so what held in one era may not hold in the next. The model was trained on 2010–2022 and tested on 2023–2026, and 2026 brought another set of new rules. The jump in pit-lane times in 2013–2014 (2.2) is still unexplained.
- **Small samples.** Monaco has 72 races (2.1), most driver-track pairs have only 5–7 starts (2.4), and the test set is 89 races. Decades are only a rough cut for eras (1.3).
- **Many tests.** With several shuffle tests, one can pass the 5% cutoff by chance. The home qualifying boost (3%) sits close to that line.
- **A simple model.** Linear regression adds up the features with fixed weights, so it can't learn that something matters more at some tracks than at others.

4.3 Next steps

- **Predict retirements too.** A second model for "will they finish?" could be combined with this one to predict the whole result.
- **Use qualifying times.** `qualifying.csv` has lap times back to 1994. The gap to pole in seconds says more than the grid slot alone, and might help at the front of the race.
- **Use lap times** (`lap_times.csv`, 1996–2026) to look at race pace and overtaking.
- **Try a model that can combine features,** such as a random forest, to test whether grid position matters more at some kinds of tracks.
- **Find out what changed in the pit lane in 2013–2014** (2.2).
