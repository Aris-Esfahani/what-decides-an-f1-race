# Part 4: Conclusions

The answer to the project's question, "what actually decides a race?", drawn from Parts 1–3, followed by the limitations and next steps. This notebook is text only, so there's nothing to run.

## 4.1 What actually decides a race?

Mostly where a driver starts, and the car that got them there.

- **The grid comes first.** Finishing order has always followed starting order closely: the grid-to-finish link was 0.70–0.78 before the 1990s and 0.76–0.80 since (1.4). In the model, scrambling grid position adds about 2 places of error per driver, more than twice as much as any other feature (3.5).
- **The car comes second.** Since 2000, the top team of each decade has won 47–49% of that decade's races (1.3), and the team's recent form is the model's second most useful feature (+0.76 places). The grid itself already reflects the car, since qualifying sets it.
- **The driver matters, but less than the car.** Once the grid and the team are known, a driver's own recent form adds a little (+0.15 places). The split is rough, though: team form and driver form move closely together, and retraining without each one costs 0.07 places for team form and 0.03 for driver form (3.5). The car still comes out ahead, but by less than the first test suggests. Drivers do differ: Verstappen turned 78% of his poles into wins, Hamilton and Schumacher about 58% (1.1).
- **Reliability used to decide races; now it rarely does.** Almost half of all starters retired in the 1950s, against 12% in the 2020s (1.2). With fewer cars breaking down, fewer drivers move up just by keeping going, which helps explain why finishers now end up closer to where they started (1.4).
- **The popular stories barely matter.** Grid position isn't more important at Monaco (2.1), pit stops haven't gotten faster overall (2.2), the home crowd helps in qualifying but not in the race (2.3), and there's no clear sign of track specialists (2.4). In the model, home races and track history add nothing (3.5). Street tracks couldn't be tested that way, because a track's type is the same for every driver in the race.

Where the answer runs out is the front of the race. The full model predicts the finishing order better than the grid alone (2.28 places off against 2.49, and a shuffle test says that isn't luck), and the version with only grid, team form and driver form does as well (2.25). But the edge shrank every season, from 0.50 places per driver in 2023 to 0.03 in 2026, and 2025 and 2026 on their own could be luck (3.4). Almost all of that gain comes from drivers starting 7th or lower. For the top 6 the model is no better than the grid: its pick won 56 of the 89 races in 2023–2026, against 54 for the pole sitter. Who wins among the front-runners probably comes down to things this data doesn't capture, like race strategy, tyres, safety cars and incidents.

## 4.2 Limitations

- **Retirements are left out of the model.** It predicts where a driver finishes if they finish. In 2023–2026 about 12% of starters didn't finish, and the model can't say who.
- **Car and driver can't be pulled apart.** The grid, team form and driver form all mix the two, so "the car comes second" is a reading of the results, not a measurement. 2.4 had the same problem with track specialists.
- **Data gaps.** Pit-stop data only covers 2011–2026 and measures the whole trip through the pit lane, not the tyre change (2.2). Lap times only go back to 1996 and weren't used. There's no weather, tyre or safety car data.
- **Rule changes.** The sport changes its rules every few years, so what held in one era may not hold in the next. The model was trained on 2010–2022 and tested on 2023–2026, and 2026 brought another set of new rules. The jump in pit-lane times in 2013–2014 (2.2) is still unexplained.
- **Small samples.** Monaco has 72 races (2.1), most driver-track pairs have only 5–7 starts (2.4), and the test set is 89 races. Decades are only a rough cut for eras (1.3).
- **Many tests.** With several shuffle tests, one can pass the 5% cutoff by chance. The home qualifying boost (2%) sits close to that line.
- **A simple model.** Linear regression adds up the features with fixed weights, so it can't learn that something matters more at some tracks than at others.

## 4.3 Next steps

- **Use qualifying times.** `qualifying.csv` has lap times back to 1994. The gap to pole in seconds says more than the grid slot alone, and might help at the front of the race.
